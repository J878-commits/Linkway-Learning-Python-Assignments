**Q1.** Convert `quantities_list = [2, 5, 1, 4, 3, 6, 2, 8]` and `prices_list = [1500, 600, 55000, 2500, 12000, 450, 18000, 300]` into NumPy arrays `qty` and `price`.
Print their `.shape`, `.dtype`, and calculate the element-wise `gross_rev = qty * price`.

```python
# ✍️ Your answer
# qty shape: (8,)
# price shape: (8,)
# qty dtype: int64
# price dtype: int64
# Gross Revenue: [ 3000  3000 55000 10000 36000  2700 36000  2400]
```

```python
# ✍️ Your code here
quantities_list = [2, 5, 1, 4, 3, 6, 2, 8]
prices_list = [1500, 600, 55000, 2500, 12000, 450, 18000, 300]

qty = np.array(quantities_list)
price = np.array(prices_list)

print("qty shape:", qty.shape)
print("price shape:", price.shape)

print("qty dtype:", qty.dtype)
print("price dtype:", price.dtype)

gross_rev = qty * price

print("Gross Revenue:", gross_rev)
```

**Q4.** Given 12 months of store sales (in ₹ thousands):
`monthly = np.array([110, 125, 140, 132, 150, 165, 158, 172, 190, 245, 270, 220])`
Slice and print:

3. Every alternate month starting from January (`monthly[::2]`)

```python
# ✍️ Your answer
# H1 Sales: [110 125 140 132 150 165]
# H1 Mean: 137.0
# H2 Sales: [158 172 190 245 270 220]
# H2 Mean: 209.16666666666666
# Alternate Months: [110 140 150 158 190 270]
```

## Part B — Boolean Filtering & `np.where` (Q6–Q10)

**Q6.** Given 10 order amounts:
`orders = np.array([1800, 12500, 4200, 28000, 950, 15600, 7400, 34000, 2100, 11200])`
Using a boolean mask `orders >= 10000`:
1. Print all orders that are `>= 10000`
2. Print the **count** of such orders (`mask.sum()`)
3. Print the **percentage** of orders that are `>= 10000` (`mask.mean() * 100`)

```python
# ✍️ Your answer
Orders >= 10000: [12500 28000 15600 34000 11200]
Count: 5
Percentage: 50.0
```

```python
# ✍️ Your code here
orders = np.array([1800, 12500, 4200, 28000, 950, 15600, 7400, 34000, 2100, 11200])

mask = orders >= 10000

print("Orders >= 10000:", orders[mask])
print("Count:", mask.sum())
print("Percentage:", mask.mean() * 100)
```

**Q9.** Shipping fee rule: Orders `< 2000` pay a `₹150` delivery fee; orders `>= 2000` get `0` delivery fee (free shipping).

**Q10.** Given an array of customer city codes from 12 orders:
`cities = np.array(["BLR", "MUM", "DEL", "BLR", "BLR", "MUM", "HYD", "BLR", "DEL", "MUM", "BLR", "DEL"])`
Use `np.unique(cities, return_counts=True)` to print each unique city, its order count, and identify the **mode** (most frequent city).

```python
# ✍️ Your answer
BLR : 5
DEL : 3
HYD : 1
MUM : 3
# Mode City: BLR
```

```python
# ✍️ Your code here
cities = np.array(["BLR", "MUM", "DEL", "BLR", "BLR", "MUM",
                   "HYD", "BLR", "DEL", "MUM", "BLR", "DEL"])

unique_cities, counts = np.unique(cities, return_counts=True)

for city, count in zip(unique_cities, counts):
    print(city, ":", count)

mode_city = unique_cities[np.argmax(counts)]

print("Mode City:", mode_city)
```

Run the cell below to create `sales_matrix`, a `4 × 6` array of monthly sales (in ₹ lakhs) for **4 regions (rows)** across **6 months Jan–Jun (columns)**.

**Q14.** An array of daily store footfall has two missing sensor readings (`np.nan`):
`footfall = np.array([320, 410, np.nan, 380, 450, np.nan, 390, 460])`
1. Count how many values are missing using `np.isnan(footfall).sum()`.
2. Calculate the **mean** and **median** footfall ignoring `NaN` using `np.nanmean()` and `np.nanmedian()`.

```python
# ✍️ Your answer
Missing Values: 2
Mean Footfall: 401.67
Median Footfall: 400.0
```

```python
# ✍️ Your code here
footfall = np.array([320, 410, np.nan, 380, 450, np.nan, 390, 460])

# 1. Count missing values
print("Missing Values:", np.isnan(footfall).sum())

# 2. Mean and Median ignoring NaN
print("Mean Footfall:", round(np.nanmean(footfall), 2))
print("Median Footfall:", np.nanmedian(footfall))
```

**Q15.** Replace the missing (`NaN`) values in `footfall` from Q14 with the **median** footfall (`np.nanmedian(footfall)`) using `np.where(np.isnan(footfall), ..., footfall)` and print the cleaned array.

```python
# ✍️ Your answer
Median Footfall: 400.0
Cleaned Footfall: [320. 410. 400. 380. 450. 400. 390. 460.]
```

```python
# ✍️ Your code here
footfall = np.array([320, 410, np.nan, 380, 450, np.nan, 390, 460])

median_footfall = np.nanmedian(footfall)

cleaned_footfall = np.where(
    np.isnan(footfall),
    median_footfall,
    footfall
)

print("Median Footfall:", median_footfall)
print("Cleaned Footfall:", cleaned_footfall)
```
